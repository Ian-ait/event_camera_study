```python
import torch
torch.cuda.device('cuda')
torch.cuda.device_count()
```

```python
def try_gpu(i = 0):
    if torch.cuda.device_count() >= i+1:
        return torch.device(f'cuda:{i}')
    return torch.device('cpu')
```

```python
x = torch.arange(9,device = try_gpu()).reshape((3,3))
x.device
```

```python
Y = torch.rand((3,3), device = try_gpu())
Y + x
```
