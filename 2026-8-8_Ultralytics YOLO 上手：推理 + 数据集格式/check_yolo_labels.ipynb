```python
from pathlib import Path
from PIL import Image, ImageDraw
import random
```

```python
ROOT = Path.cwd()
IMAGE_DIR = ROOT/"yolo_dataset"/"images"/"val"
LABEL_DIR =  ROOT/"yolo_dataset"/"labels"/"val"
OUTPUT_DIR = ROOT/"check_results"

OUTPUT_DIR.mkdir(parents = True, exist_ok= True)

images = list(IMAGE_DIR.glob('*.jpg'))
```

```python
print(images)
```

```python
for image_path in random.sample(images, 5):
    image = Image.open(image_path).convert("RGB")
    width, height = image.size
    draw = ImageDraw.Draw(image)
    label_path = LABEL_DIR/f"{image_path.stem}.txt"

    for line in label_path.read_text().splitlines():
        class_id, cx, cy, w, h = map(float, line.split())
        cx *= width
        cy *= height
        w *= width
        h *= height

        box = [cx-w/2, cy-h/2, cx+w/2, cy+h/2]

        draw.rectangle (box, outline = 'red', width = 3)
        draw.text((box[0], box[1]), str(int(class_id)), fill = "red")
    image.save(OUTPUT_DIR/image_path.name)
```
